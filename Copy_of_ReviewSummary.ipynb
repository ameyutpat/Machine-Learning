```python
import cohere
from textblob import TextBlob


# Initialize Cohere API client
cohere_client = cohere.Client(COHERE_API_KEY)

def analyze_sentiment(review):
    """Analyzes sentiment polarity (-1 to 1) and returns a label."""
    polarity = TextBlob(review).sentiment.polarity
    return "Positive" if polarity > 0 else "Negative" if polarity < 0 else "Neutral"

def summarize_reviews(reviews):
    """Generates a concise summary from multiple reviews using Cohere."""
    prompt = f"Summarize the following customer reviews into one short paragraph:\n\n{reviews}"

    try:
        response = cohere_client.generate(
            model="command-r-plus",  # Cohere's best generative model
            prompt=prompt,
            max_tokens=150,
            temperature=0.7
        )
        return response.generations[0].text.strip()
    except Exception as e:
        return f"Error generating summary: {e}"
```

```python
import pandas as pd

if __name__ == "__main__":
    # Example Amazon product ID
    product_id = "B07FZ8S74R"

    # Sample reviews
    reviews = [
        "Amazing battery life! Lasts forever.",
        "Good performance but a bit pricey.",
        "Excellent value for money. Highly recommend!",
    ]

    # Analyze sentiment for each review
    review_data = [{"review": r, "sentiment": analyze_sentiment(r)} for r in reviews]

    # Convert to DataFrame
    df = pd.DataFrame(review_data)

    # Generate AI-powered summary using Cohere
    summary = summarize_reviews("\n".join(df["review"]))

    # Print results
    print("🔹 Product ID:", product_id)
    print("\n📌 AI-Generated Review Summary:\n", summary)
    print("\n📝 Sentiment Analysis:")
    print(df)
```
